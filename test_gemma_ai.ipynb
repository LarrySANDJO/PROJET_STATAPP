```python
from mistralai import Mistral

# Initialisation
client = Mistral(api_key="dJRikqTHAfXagIP0C8nFWb2AJf7JYvz0")

# Envoi d'une requête
response = client.chat.complete(
    model="mistral-large-latest",
    messages=[
        {
            "role": "user",
            "content": "Quels sont les avantages de Mistral AI ?"
        }
    ],
    temperature=0.7,
    max_tokens=50000
)

# Affichage de la réponse
print(response.choices[0].message.content)
```
